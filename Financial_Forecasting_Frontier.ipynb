# **Project Summary -**

Banks generate vast volumes of customer, transaction and campaign data, and its value depends on how quickly it can be turned into decisions. Financial Forecasting Frontier: Distributed ML shows how distributed computing and machine learning convert banking data into actionable insight, using the bank marketing dataset (bank.csv) to simulate a real-world banking data environment.

The dataset contains 4,521 client records with 17 attributes covering demographics, financial position (balance, loans, credit default) and campaign history. The target variable, y, records whether a client subscribed to a term deposit.

The project follows the full analytics lifecycle. Data is first ingested into Hadoop and exposed through Hive tables, allowing SQL-style queries such as subscription rates by job, month or education level. This separates storage and querying from computation, as in real bank data repositories.

Apache Spark then performs exploratory analysis and preprocessing. Key findings include a heavily imbalanced target (only about 11.5% of clients subscribed), many clients never contacted before (pdays = -1 for about 82%), "unknown" categories in several fields, and strongly skewed balance and duration values. These shape the preprocessing steps, packaged in a reusable Spark ML Pipeline.

Predictive models built with Spark ML, such as Logistic Regression, Random Forest and Gradient-Boosted Trees, predict term-deposit subscription. They are compared using metrics suited to imbalanced data (AUC, precision-recall, recall, F1) with cross-validation. Because call duration is only known after a call ends, it is treated as target leakage, and models are evaluated with and without it. The business value is sharper targeting, which reduces wasted contacts and improves conversion.

A Spark Streaming component replays records as a simulated live feed, applying windowed aggregations and rule-based alerts to illustrate real-time monitoring and fraud detection. Finally, data parallelism techniques such as partitioning, caching and optimised transformations are benchmarked to show how workloads scale across executors.

Overall, the project demonstrates why distributed tools are essential in modern analytics: Hadoop and Hive provide scalable storage and querying, Spark delivers fast batch analysis and machine learning, and streaming provides timely insight. The design extends naturally to production through cloud clusters, stronger streaming pipelines and further model tuning.

````python
# 4. Bivariate Bar Plot - Subscription Rate by Marital Status
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate subscription rate for each marital category using columns from one-hot encoded df
marital_cols = [col for col in df.columns if 'marital_' in col]

# We reconstruct a temporary marital column for simple clean plotting
def get_marital(row):
    if row.get('marital_married', False):
        return 'married'
    elif row.get('marital_single', False):
        return 'single'
    else:
        return 'divorced'

temp_df = df.copy()
temp_df['marital_status'] = temp_df.apply(get_marital, axis=1)

plt.figure(figsize=(7, 5))
sns.barplot(data=temp_df, x='marital_status', y='y_yes', estimator=lambda x: sum(x)/len(x), errorbar=None, palette='pastel')
plt.title('Subscription Rate by Marital Status')
plt.xlabel('Marital Status')
plt.ylabel('Subscription Rate (Proportion)')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
````

````python
# 5. Bivariate Bar Plot - Subscription Rate by Job Category
import matplotlib.pyplot as plt
import seaborn as sns

# Identify job columns in the one-hot encoded dataset
job_cols = [col for col in df.columns if 'job_' in col]

# Map one-hot columns back to a single categorical column for plotting
def get_job(row):
    for col in job_cols:
        if row[col]:
            return col.replace('job_', '')
    return 'admin.'  # default baseline if no flag is set (due to drop_first=True)

temp_job_df = df.copy()
temp_job_df['job_type'] = temp_job_df.apply(get_job, axis=1)

# Calculate mean subscription rates and sort for better visual representation
job_order = temp_job_df.groupby('job_type')['y_yes'].mean().sort_values(ascending=False).index

plt.figure(figsize=(12, 6))
sns.barplot(data=temp_job_df, x='job_type', y='y_yes', order=job_order, errorbar=None, palette='viridis', hue='job_type', legend=False)
plt.title('Subscription Rate by Job Category')
plt.xlabel('Job Category')
plt.ylabel('Subscription Rate (Proportion)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

I chose a **Bivariate Bar Plot sorted by rate** because it is the most intuitive visual to compare conversion rates (proportions of `y_yes`) across multiple distinct categorical values (the 12 unique job sectors).

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

### **Hypothesis 1: Account Balance and Term Deposit Subscriptions**

* **Null Hypothesis ($H_0$):** There is no significant difference in the mean average yearly account balance between bank clients who subscribed to a term deposit and those who did not subscribe.
  $$H_0: \mu_{\text{subscriber}} = \mu_{\text{non-subscriber}}$$

* **Alternative Hypothesis ($H_a$):** Bank clients who subscribed to a term deposit have a significantly different (or higher) mean average yearly account balance compared to those who did not subscribe.
  $$H_a: \mu_{\text{subscriber}} \neq \mu_{\text{non-subscriber}}$$


#### 2. Perform an appropriate statistical test.

````python
from scipy import stats

# Split the dataset into subscribers and non-subscribers groups
subscribers_balance = df[df['y_yes'] == True]['balance']
non_subscribers_balance = df[df['y_yes'] == False]['balance']

# Display basic descriptive statistics
print(f"Subscribers Mean Balance: ${subscribers_balance.mean():.2f} (Std: {subscribers_balance.std():.2f}, Count: {len(subscribers_balance)})")
print(f"Non-subscribers Mean Balance: ${non_subscribers_balance.mean():.2f} (Std: {non_subscribers_balance.std():.2f}, Count: {len(non_subscribers_balance)})")

# Perform Welch's t-test (equal_var=False because standard deviations and group sizes differ significantly)
t_stat, p_val = stats.ttest_ind(subscribers_balance, non_subscribers_balance, equal_var=False)

print(f"\nT-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_val:.4e}")
````

##### Which statistical test have you done to obtain P-Value?

We performed a **Welch's Two-Sample Independent T-test** (using `scipy.stats.ttest_ind` with `equal_var=False`).

### **Results Summary:**
* **Subscriber Mean Balance:** $1571.96 ($s.d. = 2444.40$)
* **Non-Subscriber Mean Balance:** $1403.21 ($s.d. = 3075.35$)
* **T-Statistic:** 1.4347
* **P-Value:** $1.5178 \times 10^{-1}$ (0.1518)

### **Statistical Conclusion:**
Using a standard significance level of $\alpha = 0.05$, we find that the $p$-value ($0.1518$) is **greater than** $0.05$. Therefore, we **fail to reject the null hypothesis ($H_0$)**. This indicates that the observed difference in mean balances between subscribers and non-subscribers is not statistically significant at a 95% confidence level.


##### Why did you choose the specific statistical test?

We chose **Welch's T-Test** for the following reasons:

1. **Continuous vs. Categorical Variables:** We are comparing a continuous numerical variable (`balance`) across two groups defined by a binary categorical target variable (`y_yes`).
2. **Unequal Group Sizes:** The groups are highly unbalanced, with 4,000 non-subscribers and only 521 subscribers.
3. **Unequal Variances (Heteroscedasticity):** The standard deviation of the two groups differs significantly ($3075.35$ vs. $2444.40$). Welch's t-test adjusts the degrees of freedom to remain robust against unequal variances, making it far more reliable than the standard Student's t-test.

````python
# Checking for missing values again before starting pre-processing
missing_values = df.isnull().sum()
print("Missing values per column in current DataFrame:")
print(missing_values[missing_values > 0] if (missing_values > 0).any() else "No missing values found!")

# Although no null values are present in this cleaned dataset, a production-ready approach
# is to implement a robust safety step for missing value imputation.
# We define a function to impute any unexpected future missing values:
def impute_missing_values(data):
    df_imputed = data.copy()
    for col in df_imputed.columns:
        if df_imputed[col].isnull().any():
            if df_imputed[col].dtype in ['int64', 'float64']:
                # Impute numeric features with median
                median_val = df_imputed[col].median()
                df_imputed[col].fillna(median_val, inplace=True)
            else:
                # Impute categorical features with mode
                mode_val = df_imputed[col].mode()[0]
                df_imputed[col].fillna(mode_val, inplace=True)
    return df_imputed

df = impute_missing_values(df)
print("\nMissing value handling/imputation pipeline executed successfully!")
````

````python
# Detect and treat outliers using the IQR (Interquartile Range) method
# We will cap extreme values at the 1.5 * IQR boundaries for key continuous numerical features

# Keep a copy of raw values before treatment for verification
df_before_treatment = df.copy()

columns_to_treat = ['balance', 'duration', 'age']
print("Before outlier treatment stats:")
display(df[columns_to_treat].describe())

for col in columns_to_treat:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # Cap outliers at lower and upper boundaries (Winsorization/Capping technique)
    df[col] = np.clip(df[col], lower_limit, upper_limit)

print("\nAfter outlier treatment (Capping/Winsorization) stats:")
display(df[columns_to_treat].describe())
````

   * **What it does:** Since account balances can contain negative values (due to overdrafts, with our capped minimum being $-2047.5$), a simple logarithm cannot be computed directly because $\log(x)$ is undefined for $x \le 0$.
   * **Why we used it:** We dynamically calculated a positive offset shift (`shift_val = abs(min_balance) + 1`), added it to the balance vector to ensure all values were positive, and then safely applied `log1p`.

````python
# Apply mathematical data transformations to continuous features that are still heavily skewed
# We use log1p (log(1 + x)) to handle zeros and shift skewed values into normal-like curves

# Keep a copy for comparative visualization or model testing
df_transformed = df.copy()

# Let's transform balance (requires shift adjustment because it has negative capped values)
# Shift balance so all values are positive before applying log1p
shift_val = abs(df['balance'].min()) + 1
df['transformed_balance'] = np.log1p(df['balance'] + shift_val)

# Transform duration
df['transformed_duration'] = np.log1p(df['duration'])

# Transform age
df['transformed_age'] = np.log1p(df['age'])

# Verify transformed distributions
print("Original skewness values:")
print(df_transformed[['balance', 'duration', 'age']].skew())

print("\nTransformed skewness values (approaching zero/normal):")
print(df[['transformed_balance', 'transformed_duration', 'transformed_age']].skew())

# Drop original continuous features to prevent redundancy
df.drop(columns=['balance', 'duration', 'age'], inplace=True)
print("\nNumerical attributes transformed successfully and original features retired.")
````

* **Split Representation:** In our training split of 3,616 samples, only 417 are positive subscription cases.
* **Consequences of Imbalance:** If we train a standard machine learning classifier on this data without handling the imbalance, the model will optimize for overall accuracy. It could simply predict "no subscription" for every single client, achieving an empty yet mathematically high accuracy of 88.5% while completely failing to identify actual potential subscribers. This makes handling the imbalance critical.

````python
# Handling Imbalanced Dataset using SMOTE (Synthetic Minority Over-sampling Technique)
from imblearn.over_sampling import SMOTE

print(f"Target class balance before SMOTE in training set:")
print(y_train.value_counts())

# Initialize SMOTE
smote = SMOTE(random_state=42)

# Apply SMOTE to the training data only to prevent data leakage
X_train_balanced, y_train_balanced = smote.fit_resample(X_train, y_train)

print(f"\nTarget class balance after SMOTE in training set:")
print(y_train_balanced.value_counts())
print(f"\nBalanced training set shape: {X_train_balanced.shape}")
````

##### What technique did you use to handle the imbalance dataset and why? (If needed to be balanced)

### **What technique did you use to handle the imbalance dataset and why?**

I used **SMOTE (Synthetic Minority Over-sampling Technique)** to handle the dataset imbalance.

### **Justification:**
1. **Avoids Information Loss:** Unlike random under-sampling (which discards majority class instances and reduces our already small dataset), SMOTE preserves all available majority class records.
2. **Prevents Overfitting:** Instead of duplicating exact minority class records (which leads to severe overfitting), SMOTE synthesizes new, realistic, non-identical minority instances by interpolating between existing positive class neighbors in the feature space.
3. **Applied Only to Training Set:** SMOTE was strictly applied to `X_train` and `y_train`. The test set (`X_test` and `y_test`) was left completely untouched to ensure our model is evaluated on realistic, real-world imbalanced data distributions.

#### 2. Cross- Validation & Hyperparameter Tuning

````python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

# Define the parameter grid for Logistic Regression tuning
param_grid = {
    'C': [0.001, 0.01, 0.1, 1, 10, 100],
    'penalty': ['l1', 'l2'],
    'solver': ['saga']  # 'saga' supports both l1 and l2 regularization
}

# Initialize baseline Logistic Regression
log_reg_base = LogisticRegression(random_state=42, max_iter=2000)

# Configure GridSearchCV with 5-fold Stratified Cross-Validation
grid_search = GridSearchCV(
    estimator=log_reg_base,
    param_grid=param_grid,
    scoring='roc_auc',
    cv=5,
    n_jobs=-1,
    verbose=1
)

# Fit the grid search on our balanced training set
grid_search.fit(X_train_balanced, y_train_balanced)

# Output best parameters and cross-validation score
print(f"Best Hyperparameters: {grid_search.best_params_}")
print(f"Best Cross-Validation ROC-AUC: {grid_search.best_score_:.4f}")

# Evaluate the optimized model on the untouched test set
best_log_reg = grid_search.best_estimator_
y_pred_tuned = best_log_reg.predict(X_test)
y_pred_proba_tuned = best_log_reg.predict_proba(X_test)[:, 1]

# Display tuned performance metrics
print("\n=== Tuned Logistic Regression Performance on Test Set ===")
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_tuned))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_tuned))

tuned_auc = roc_auc_score(y_test, y_pred_proba_tuned)
print(f"Tuned ROC-AUC Score: {tuned_auc:.4f}")
````

##### Which hyperparameter optimization technique have you used and why?

### **Which hyperparameter optimization technique have you used and why?**

We utilized **GridSearchCV (Grid Search Cross-Validation)** combined with **5-Fold Stratified Cross-Validation**:

1. **Comprehensive Search**: GridSearchCV systematically evaluates all possible parameter combinations from the grid (`C` and `penalty`). This guarantees that we locate the absolute global optimum within our defined search space.
2. **Prevention of Overfitting**: By performing 5-Fold Stratified Cross-Validation on the balanced training set, we ensure the selected hyperparameters generalize well across different random subsets of the dataset.
3. **Stratification**: Stratified folds keep the proportion of target classes identical across all splits, preventing biased parameter selection on imbalanced data.
4. **Feature Selection with L1 Penalty**: Utilizing `penalty=['l1', 'l2']` with the `saga` solver allows the algorithm to perform automatic feature selection via Lasso (L1) regularization by shrinking non-essential feature coefficients to exactly zero, resulting in a cleaner and highly interpretable model.
